```python
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax1.scatter(x,y,c="red")
ax2.bar(x,y)
ax3.hist(x)
ax4.barh(x,y)
```

```python
heart_disease=pd.read_csv("heart-disease (1).csv")
heart_disease
```

```python
over_50=heart_disease[heart_disease["age"]>50]
over_50
```

```python
over_50.plot(x="age",
            y="chol",
            kind="scatter",
            c="target",
            cmap="winter",
            figsize=(10,5))
```

```python
fig,ax=plt.subplots(figsize=(10,5))
over_50.plot(x="age",
            y="chol",
            kind="scatter",
            c="target",
            cmap="winter",
            ax=ax);
#ax.set_xlim(50,65)
```

```python
over_50
```

```python
plt.style.use("seaborn-v0_8-darkgrid")
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,7))
scatter=ax[0].scatter(over_50["age"],
             over_50["chol"],
              c=over_50["target"],
              cmap="winter")
ax[0].set(title="Heart Disease and Cholestrol level",
        xlim=[50,80], ylabel="Cholestrol")
ax[0].legend(*scatter.legend_elements(),
            title="Target");

ax[0].axhline(y=over_50["chol"].mean(),
             linestyle="--",
             color="r",
             label="Average")

scatter1=ax[1].scatter(over_50["age"],
             over_50["thalach"],
              c=over_50["target"],
              cmap="winter")
ax[1].set(ylabel="Maximum heart rate")
ax[1].legend(*scatter1.legend_elements(),
            title="Target");

ax[1].axhline(y=over_50["thalach"].mean(),
             linestyle="--",
             color="r",
             label="Average")
fig.suptitle("Heart Disease Analysis",
            fontweight="bold",
            fontsize="15")
```
